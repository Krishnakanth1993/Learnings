```python
import random
import numpy as np
```

```python
text = ['How are you doing today?', 
'I hope you are having a great day!', 
'What are your plans for the weekend?'
]
```

```python
allwords = [x.lower() for line in text for x in line.split()]
vocab = sorted(set(allwords))
```

```python
print(allwords)
print(vocab)
```

```python
word2idx = {w: i for i, w in enumerate(vocab)}
print(word2idx)
```

```python
idx2word = {i: w for i, w in enumerate(vocab)}
print(idx2word)
```

```python
def encode(text):
    return [word2idx[w] for w in text.split()]

def decode(indices):
    return ' '.join([idx2word[i] for i in indices])
```

```python
text = 'How are you doing today?'
encoded = encode(text)
print('Encoded:', encoded)
```

```python
decoded = decode(encoded)
print('Decoded:', decoded)
```

```python
assert decode(encoded) == text
```

```python
random_indices = random.choices(range(len(vocab)), k=5)
print('Random Indices:', random_indices)
```

```python
decoded = decode(random_indices)
print('Decoded:', decoded)
```

```python
search_word = 'you'
id_list = []
for i , word in enumerate(allwords):
    context = []
    context_token_ids = []
    if word == search_word.lower():
        print(f'Index of the word "{search_word}": {i}')
        id_list.append(i)
        context = allwords[max(0, i-1): i+2]
        print(f'Context words: {context}')
        context_token_ids = [word2idx[w] for w in context]
        print (f'Context token IDs: {context_token_ids}')
print(f'All indices of the word "{search_word}": {id_list}')  
```

```python
search_word = 'are'
id_list = [i for i, word in enumerate(allwords) if word == search_word.lower()]
print(f'All indices of the word "{search_word}": {id_list}')

for i in id_list:
    context = allwords[max(0, i-1): i+2]
    print(f'Context words: {context}')
    print(' '.join(context))
    context_token_ids = [word2idx[w] for w in context]
    print (f'Context token IDs: {context_token_ids}')
```
